```python
from soynlp.tokenizer import LTokenizer
import re

# LTokenizer를 이용한 형태소 분석
tokenizer = LTokenizer()



# 조사 패턴을 정규식으로 정의 (좀 더 포괄적으로 적용)
josa_pattern = re.compile(r'\b[가-힣]+(은|는|이|가|을|를|에|의|도|와|과|으로|에서|한테|에게|부터|까지|만)\b')

# 텍스트 전처리 함수
def preprocess_text(text):
    # 조사 제거 (정규 표현식을 통해 조사와 붙은 형태의 단어도 처리)
    text = josa_pattern.sub(lambda x: x.group(0)[:-1], text)
    
    # 특수문자 제거
    text = re.sub(r'[^가-힣\s]', '', text)
    
    # 형태소 분석 및 불용어 제거
    tokens = tokenizer.tokenize(text)
    tokens = [word for word in tokens if word not in stopwords and len(word) > 1]
    
    # 다시 문자열로 결합
    return ' '.join(tokens)

# 전처리 적용
processed_data = [preprocess_text(sentence) for sentence in numbered_data.values()]

# 결과 출력
for i, text in enumerate(processed_data):
    original_text = list(numbered_data.values())[i]  # 원본 텍스트 가져오기
    print(f"Original: {original_text}")
    print(f"Processed: {text}")
    print('---')
```

```python
# 문장 리스트
sentences = list(processed_data)

# 각 문장의 임베딩 구하기
def get_embeddings(sentences):
    embeddings = []
    for sentence in sentences:
        inputs = tokenizer(sentence, return_tensors='pt', padding=True, truncation=True).to(device)
        with torch.no_grad():
            outputs = model(**inputs)
            sentence_embedding = outputs.last_hidden_state.mean(dim=1).cpu().numpy()
            embeddings.append(sentence_embedding)
    return np.vstack(embeddings)

# 문장 임베딩
sentence_embeddings = get_embeddings(sentences)
print(f"Sentence Embeddings Shape: {sentence_embeddings.shape}")
```

```python
# TF-IDF 벡터
from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF 벡터화
def get_tfidf_features(sentences):
    vectorizer = TfidfVectorizer(max_features=128)
    tfidf_matrix = vectorizer.fit_transform(sentences).toarray()
    return tfidf_matrix

# TF-IDF 피처 생성
tfidf_features = get_tfidf_features(sentences)
print(f"TF-IDF Features Shape: {tfidf_features.shape}")
```

```python
# BERT 임베딩, TF-IDF, 감정 분석
combined_features = np.hstack([tfidf_features])

# 결합된 피처 출력
print(f"Combined Features Shape: {combined_features.shape}")
```

```python
# Manhattan Similarity Matrix 계산 함수
def manhattan_similarity_matrix(embeddings):
    num_sentences = embeddings.shape[0]
    similarity_matrix = np.zeros((num_sentences, num_sentences))

    for i in range(num_sentences):
        for j in range(num_sentences):
            similarity_matrix[i, j] = np.sum(np.abs(embeddings[i] - embeddings[j]))  # Manhattan 거리 계산

    return similarity_matrix
```

```python
# Manhattan 유사도 행렬 계산
manhattan_matrix = manhattan_similarity_matrix(combined_features)
print(f"Manhattan Similarity Matrix:\n{manhattan_matrix}")
```
