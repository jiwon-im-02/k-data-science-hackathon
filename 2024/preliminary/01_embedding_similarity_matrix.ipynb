```python
import torch
from transformers import BertModel, BertTokenizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

from transformers import BertTokenizer
from kobert_transformers import get_tokenizer
```

```python
tokenizer = get_tokenizer()
model = BertModel.from_pretrained('skt/kobert-base-v1')

# GPU 사용 가능 시 설정 (없으면 CPU 사용)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

# 아래 뜨는거 미래 경고임 없애고 싶으면 torenizer 함수 쓸 때, clean_up_tokenization_spaces=False 인수 추가  
```

```python
# 문장 리스트
sentences = list(numbered_data.values())

# 각 문장의 임베딩 구하기
def get_embeddings(sentences):
    embeddings = []
    for sentence in sentences:
        inputs = tokenizer(sentence, return_tensors='pt', padding=True, truncation=True).to(device)
        with torch.no_grad():
            outputs = model(**inputs)
            sentence_embedding = outputs.last_hidden_state.mean(dim=1).cpu().numpy()
            embeddings.append(sentence_embedding)
    return np.vstack(embeddings)

# 문장 임베딩
sentence_embeddings = get_embeddings(sentences)
print(f"Sentence Embeddings Shape: {sentence_embeddings.shape}")
```

```python
# TF-IDF 벡터
from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF 벡터화
def get_tfidf_features(sentences):
    vectorizer = TfidfVectorizer(max_features=1000)
    tfidf_matrix = vectorizer.fit_transform(sentences).toarray()
    return tfidf_matrix

# TF-IDF 피처 생성
tfidf_features = get_tfidf_features(sentences)
print(f"TF-IDF Features Shape: {tfidf_features.shape}")
```

```python
# 감정 분석
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# 감정 분석
def get_sentiment_scores(sentences):
    analyzer = SentimentIntensityAnalyzer()
    sentiment_features = []
    for sentence in sentences:
        scores = analyzer.polarity_scores(sentence)
        sentiment_features.append([scores['neg'], scores['neu'], scores['pos'], scores['compound']])
    return np.array(sentiment_features)

# 감정 분석 피처 생성
sentiment_features = get_sentiment_scores(sentences)
print(f"Sentiment Features Shape: {sentiment_features.shape}")
```

```python
# BERT 임베딩, TF-IDF, 감정 분석
combined_features = np.hstack([sentence_embeddings, tfidf_features, sentiment_features])

# 결합된 피처 출력
print(f"Combined Features Shape: {combined_features.shape}")
```

```python
# Manhattan Similarity Matrix 계산 함수
def manhattan_similarity_matrix(embeddings):
    num_sentences = embeddings.shape[0]
    similarity_matrix = np.zeros((num_sentences, num_sentences))

    for i in range(num_sentences):
        for j in range(num_sentences):
            similarity_matrix[i, j] = np.sum(np.abs(embeddings[i] - embeddings[j]))  # Manhattan 거리 계산

    return similarity_matrix
```

```python
# Manhattan 유사도 행렬 계산
manhattan_matrix = manhattan_similarity_matrix(combined_features)
print(f"Manhattan Similarity Matrix:\n{manhattan_matrix}")
```
